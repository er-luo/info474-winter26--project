```python
folder  = 'data/Maust_et_al_data'

weather_data = pd.read_csv(os.path.join(folder, 'weather_data.csv'))
weather_data_og = weather_data.copy()   
stages_data = pd.read_csv(os.path.join(folder, 'stages_data.csv'))
blossom_data = pd.read_csv(os.path.join(folder, 'blossom_dates.csv'))
print("Loaded uw data files successfully.")

uw_data = []
uw_data.append(weather_data)
uw_data.append(stages_data)
uw_data.append(blossom_data)
print("uw files appended to list successfully.")  

sdot_data = pd.read_csv('data/SDOT_Trees_data.csv', low_memory=False)
print("Loaded SDOT tree data successfully.")
```

```python
def clean(df, columns_to_keep, drop_dupes_column, drop_na):    
    df.columns = df.columns.str.lower()
    df = df[columns_to_keep]
    og = df.copy()
    if drop_na:
        df = df.dropna(subset=drop_na)
    if drop_dupes_column is not None:
        df = df.drop_duplicates(subset=[drop_dupes_column])
    removed = len(og) - len(df)
    print(f"{removed} rows dropped")
    return df
```

```python
weather_data = clean(weather_data, ['date', 'tmax', 'tmin'], 'date', False)
stages_data = clean(stages_data, stages_data.columns.str.lower(), None, False)
blossom_data = clean(blossom_data, blossom_data.columns.str.lower(), None, False)

folder = 'data/clean'
weather_data.to_csv(os.path.join(folder, 'output_file.csv'), index=False)
stages_data.to_csv(os.path.join(folder, 'stages_output_file.csv'), index=False)
blossom_data.to_csv(os.path.join(folder, 'blossom_output_file.csv'), index=False)
print("Cleaned data files saved successfully.")
```
